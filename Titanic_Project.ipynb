Task 1: Loading Dataset and Basic Statistical Analysis

In this task, the Titanic dataset is loaded using Pandas. Basic information about the dataset is explored using info(), describe(), head(), and shape() methods.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("train.csv")

df.head()
```

```python
df.to_csv("titanic_cleaned.csv", index=False)
```

Task 2: Handling Missing Data

Missing values are identified and handled using Mean, Median, and Mode imputation techniques to improve data quality.

```python
# Check missing values before cleaning
print("Before Cleaning:")
print(df.isnull().sum())

# Mean imputation
df['Fare'] = df['Fare'].fillna(df['Fare'].mean())

# Median imputation
df['Age'] = df['Age'].fillna(df['Age'].median())

# Mode imputation
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("\nAfter Cleaning:")
print(df.isnull().sum())
```

Task 3: Encoding Categorical Variables

Categorical variables are converted into numerical format using Label Encoding and One-Hot Encoding.

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Label Encoding
le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])

print("Label Encoding Completed")
```

```python
# One Hot Encoding
df = pd.get_dummies(df, columns=['Embarked'], dtype=int)

print("One Hot Encoding Completed")
display(df.head())
```
